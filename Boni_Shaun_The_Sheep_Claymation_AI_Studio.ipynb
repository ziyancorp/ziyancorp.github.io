```python
# @title 🖼️ 2. Ekstrak Aset Karakter Boni & Tobi (Claymation HD)
import os, base64
os.makedirs('boni_assets', exist_ok=True)

b64_scene1 = ''''''
b64_scene2 = ''''''
b64_scene3 = ''''''

with open('boni_assets/scene1.jpg', 'wb') as f: f.write(base64.b64decode(b64_scene1))
with open('boni_assets/scene2.jpg', 'wb') as f: f.write(base64.b64decode(b64_scene2))
with open('boni_assets/scene3.jpg', 'wb') as f: f.write(base64.b64decode(b64_scene3))

print('✅ 3 Adegan Claymation Boni & Tobi Siap di Folder: boni_assets/')
```

```python
# @title 🧠 3. Muat AI Model Image-to-Video (Stable Video Diffusion XT)
import torch
from diffusers import StableVideoDiffusionPipeline
from diffusers.utils import load_image, export_to_video
from PIL import Image

print('⏳ Memuat Model SVD-XT ke VRAM GPU...')
pipe = StableVideoDiffusionPipeline.from_pretrained(
    'stabilityai/stable-video-diffusion-img2vid-xt',
    torch_dtype=torch.float16,
    variant='fp16'
)
pipe.enable_model_cpu_offload()
pipe.unet.enable_forward_chunking()
print('✅ Model SVD-XT Berhasil Dimuat di GPU!')
```

```python
# @title 🎬 4. Render Animasi Gerak Penuh Tiap Adegan (Image-to-Video AI)
os.makedirs('rendered_clips', exist_ok=True)

scenes_config = [
    {
        'name': 'scene1_motion',
        'path': 'boni_assets/scene1.jpg',
        'motion_bucket_id': 127,
        'fps': 12,
        'noise_aug': 0.05
    },
    {
        'name': 'scene2_motion',
        'path': 'boni_assets/scene2.jpg',
        'motion_bucket_id': 180,
        'fps': 12,
        'noise_aug': 0.08
    },
    {
        'name': 'scene3_motion',
        'path': 'boni_assets/scene3.jpg',
        'motion_bucket_id': 150,
        'fps': 12,
        'noise_aug': 0.06
    }
]

for sc in scenes_config:
    print(f"🎥 Merender Gerakan AI: {sc['name']}...")
    image = load_image(sc['path']).resize((576, 1024))
    generator = torch.manual_seed(42)
    frames = pipe(
        image,
        decode_chunk_size=4,
        generator=generator,
        motion_bucket_id=sc['motion_bucket_id'],
        noise_aug_strength=sc['noise_aug'],
        num_frames=25
    ).frames[0]
    
    out_path = f"rendered_clips/{sc['name']}.mp4"
    export_to_video(frames, out_path, fps=sc['fps'])
    print(f"✅ Selesai: {out_path}")

print('🎉 Seluruh Klip Animasi AI Selesai Dirender!')
```

```python
# @title 🎙️ 5. Generate Voiceover Narasi & Sound Effects (SFX Kartun)
import edge_tts, asyncio, subprocess

narrations = [
    ('voice_1.mp3', 'Di sebuah peternakan bukit hijau yang cerah, Boni si domba gembul dan sahabatnya Tobi si anjing kecil sedang asyik bersantai menikmati semilir angin sepoi-sepoi.'),
    ('voice_2.mp3', 'Tapi tiba-tiba... Gedebuk! Sebuah semangka raksasa menggelinding kencang menuruni bukit! Awas! Semangka itu meluncur lurus ke arah traktor pak tani!'),
    ('voice_3.mp3', 'Dengan sigap, Boni memanggil kawanan domba! Mereka bergotong royong menyusun benteng jerami emas... Bruuk! Semangka raksasa berhasil ditahan dengan selamat! Kerja sama yang sangat luar biasa!')
]

async def make_audio():
    for fname, text in narrations:
        comm = edge_tts.Communicate(text, 'id-ID-GadisNeural', rate='+6%', pitch='+4Hz')
        await comm.save(fname)
        print(f'✅ Voiceover dibuat: {fname}')

await make_audio()

# Generate Kartun Sound Effects & Background Music
!ffmpeg -y -f lavfi -i "anoisesrc=d=45:c=pink:r=44100:a=0.01" -f lavfi -i "sine=f=523.25:d=45" -filter_complex "[1]volume=0.08,apulsator=hz=2[mel];[0][mel]amix=inputs=2[aout]" -map "[aout]" -t 42 bgm.wav
print('✅ Audio & SFX Track Siap!')
```

```python
# @title 🎬 6. Assembling Master Video 1080p dengan Subtitle Kartun Animasi
from PIL import Image, ImageDraw, ImageFont
import subprocess

sub_texts = [
    ('PETUALANGAN BONI & TOBI', 'Di peternakan bukit hijau yang cerah...', 'Boni & Tobi sedang asyik bersantai menikmati semilir angin.'),
    ('AWAS SEMANGKA RAKSASA!', 'Tiba-tiba... GEDEEBUK! Semangka raksasa meluncur!', 'Awas! Menuju lurus ke arah traktor pak tani!'),
    ('BENTENG JERAMI EMAS!', 'Boni & kawanan domba kompak menyusun jerami...', 'BRUUK! Semangka berhasil ditahan! Kerja sama hebat!')
]

for idx, (title, l1, l2) in enumerate(sub_texts, 1):
    img = Image.new('RGBA', (1080, 1920), (0,0,0,0))
    draw = ImageDraw.Draw(img)
    try:
        f_title = ImageFont.truetype('/usr/share/fonts/truetype/dejavu/DejaVuSans-Bold.ttf', 44)
        f_sub = ImageFont.truetype('/usr/share/fonts/truetype/dejavu/DejaVuSans-Bold.ttf', 36)
    except:
        f_title = f_sub = ImageFont.load_default()
        
    draw.rounded_rectangle([200, 100, 880, 190], radius=25, fill=(30, 30, 30, 210), outline=(255, 215, 0, 255), width=3)
    draw.text((540, 145), title, fill=(255, 215, 0, 255), font=f_title, anchor='mm')
    
    draw.rounded_rectangle([40, 1620, 1040, 1820], radius=30, fill=(0, 0, 0, 220), outline=(255, 255, 255, 200), width=2)
    draw.text((540, 1675), l1, fill=(255, 235, 59, 255), font=f_sub, anchor='mm')
    draw.text((540, 1755), l2, fill=(255, 255, 255, 255), font=f_sub, anchor='mm')
    img.save(f'overlay_{idx}.png')

# Render Final Scene Clips
final_clips = []
for idx in [1, 2, 3]:
    in_mp4 = f'rendered_clips/scene{idx}_motion.mp4'
    in_voice = f'voice_{idx}.mp3'
    in_over = f'overlay_{idx}.png'
    out_sc = f'final_scene_{idx}.mp4'
    
    cmd = [
        'ffmpeg', '-y',
        '-stream_loop', '-1', '-i', in_mp4,
        '-loop', '1', '-i', in_over,
        '-i', in_voice,
        '-filter_complex', '[0:v]scale=1080:1920:force_original_aspect_ratio=increase,crop=1080:1920[bg];[bg][1:v]overlay=0:0[vout]',
        '-map', '[vout]', '-map', '2:a',
        '-c:v', 'libx264', '-pix_fmt', 'yuv420p', '-c:a', 'aac', '-shortest',
        out_sc
    ]
    subprocess.run(cmd, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
    final_clips.append(out_sc)
    print(f'✅ Final Scene {idx} Siap!')

# Concat ke Master Video
with open('concat.txt', 'w') as f:
    for c in final_clips:
        f.write(f"file '{c}'\n")

!ffmpeg -y -f concat -safe 0 -i concat.txt -i bgm.wav -filter_complex "[0:a]volume=1.0[v];[1:a]volume=0.2[bgm];[v][bgm]amix=inputs=2:duration=first[aout]" -map 0:v -map "[aout]" -c:v libx264 -pix_fmt yuv420p -c:a aac Master_Boni_Shaun_The_Sheep_Animasi_HD.mp4

print('\n🎉 MASTER VIDEO ANIMASI AI SELESAI DIRENDER: Master_Boni_Shaun_The_Sheep_Animasi_HD.mp4')

# Preview Video langsung di notebook
from IPython.display import HTML
from base64 import b64encode
mp4 = open('Master_Boni_Shaun_The_Sheep_Animasi_HD.mp4','rb').read()
data_url = "data:video/mp4;base64," + b64encode(mp4).decode()
HTML(f"""<video width=360 controls autoplay loop><source src="{data_url}" type="video/mp4"></video>""")
```
